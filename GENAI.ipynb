```python
import os
import json

# Define the base path to the brand folders
base_dir = "data_google/data_google (1)"
output_metadata = {}
output_path = "data_google/image_metadata.json"

# Define the order in which brands should be merged
brand_order = ["Maruti suzuki", "Tata", "Hyundai", "Mahindra"]

counter = 1  # start numbering from 1

print(f"🔍 Scanning directory: {base_dir}")
for brand in brand_order:
    brand_path = os.path.join(base_dir, brand)
    metadata_file = os.path.join(brand_path, "image_metadata.json")

    if os.path.exists(metadata_file):
        with open(metadata_file, "r") as f:
            brand_metadata = json.load(f)
            for _, entry in brand_metadata.items():
                output_metadata[str(counter)] = entry
                counter += 1
        print(f"✅ Merged {len(brand_metadata)} entries from {brand}")
    else:
        print(f"⚠️ Metadata not found for {brand}: {metadata_file}")

# Save merged metadata
os.makedirs(os.path.dirname(output_path), exist_ok=True)
with open(output_path, "w") as f:
    json.dump(output_metadata, f, indent=2)

print(f"\n✅ Final metadata saved to {output_path}")
print(f"📦 Total entries merged: {len(output_metadata)}")
```

```python
import os
import json
from PIL import Image
import torch
import faiss
import numpy as np
from transformers import CLIPProcessor, CLIPModel

# Load CLIP model and processor
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
model.eval()

# Load metadata from previous scraping
with open("data_google/image_metadata.json", "r", encoding="utf-8") as f:
    image_metadata_dict = json.load(f)

# Store embeddings and metadata
image_vectors = []
image_ids = []
image_id_to_meta = {}

for image_id, meta in image_metadata_dict.items():
    try:
        img_path = meta["img_path"]
        if not os.path.exists(img_path):
            print(f"⚠️ Image not found: {img_path}")
            continue

        image = Image.open(img_path).convert("RGB")
        inputs = processor(images=image, return_tensors="pt")

        with torch.no_grad():
            embedding = model.get_image_features(**inputs)

        # Normalize to unit vector
        embedding = embedding / embedding.norm(p=2, dim=-1, keepdim=True)

        image_vectors.append(embedding.squeeze().numpy())
        image_ids.append(str(image_id))
        image_id_to_meta[str(image_id)] = meta

        print(f"✅ Embedded: {img_path}")
    except Exception as e:
        print(f"❌ Failed embedding {meta['img_path']}: {e}")

# Convert to NumPy array
if not image_vectors:
    raise RuntimeError("No valid image embeddings created.")

image_vectors_np = np.vstack(image_vectors).astype("float32")

# Build FAISS index
dimension = image_vectors_np.shape[1]
faiss_index = faiss.IndexFlatL2(dimension)
faiss_index.add(image_vectors_np)

print(f"✅ FAISS index created with {faiss_index.ntotal} images.")

# Save index and metadata
faiss.write_index(faiss_index, "image_index.faiss")
with open("image_id_to_meta.json", "w", encoding="utf-8") as f:
    json.dump(image_id_to_meta, f, indent=2)

print("✅ Image index and metadata saved.")
```

```python
from llama_index.core import VectorStoreIndex, StorageContext, Settings
from llama_index.core.readers import SimpleDirectoryReader
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from pathlib import Path

# Define the directory containing description/spec text files
text_dir = Path("brand_texts")

# Load documents from the text directory
documents = SimpleDirectoryReader(str(text_dir)).load_data()

# Set up embedding model
embed_model = HuggingFaceEmbedding(model_name="all-MiniLM-L6-v2")
Settings.embed_model = embed_model

# Create and persist vector index
persist_dir = "./storage"
index = VectorStoreIndex.from_documents(documents)
index.storage_context.persist(persist_dir=persist_dir)

print("✅ Vector index created and stored successfully for car descriptions/specs.")
```

```python
import os
import json
import torch
import faiss
import numpy as np
from PIL import Image
from IPython.display import display as ipy_display
from transformers import CLIPProcessor, CLIPModel
import google.generativeai as genai
import re

from llama_index.core import StorageContext, load_index_from_storage

# 🔑 Set Gemini API Key
os.environ["GOOGLE_API_KEY"] = "  "  # enter the api key here
genai.configure(api_key=os.environ["GOOGLE_API_KEY"])

# ✅ Load CLIP model and processor
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
clip_model.eval()

# ✅ Load FAISS image index and metadata
faiss_index = faiss.read_index("image_index.faiss")
with open("image_id_to_meta.json", "r", encoding="utf-8") as f:
    image_id_to_meta = json.load(f)
image_ids = list(image_id_to_meta.keys())

# ✅ Load LlamaIndex vector index (car spec docs)
storage_context = StorageContext.from_defaults(persist_dir="./storage")
llm_index = load_index_from_storage(storage_context)
retriever = llm_index.as_retriever()

# ✅ Your user query
test_query = "images and specifications of maruthi suzuki alto K10 LXI ."

# ✅ Extract model name from query (e.g., "Aura")
match = re.search(r"hyundai\s+([a-zA-Z0-9]+)", test_query.lower())
target_model_name = match.group(1) if match else ""

# ✅ Convert text query to CLIP embedding
inputs = clip_processor(text=[test_query], return_tensors="pt", padding=True)
with torch.no_grad():
    text_features = clip_model.get_text_features(**inputs)
    text_features = text_features / text_features.norm(p=2, dim=-1, keepdim=True)
    text_vector = text_features.cpu().numpy().astype("float32")

# ✅ Search FAISS index for top-k image matches
top_k = 2
distances, indices = faiss_index.search(text_vector, top_k)

# ✅ Load and filter matching images by metadata
gemini_images = []
print(f"\n🖼️ Top Images Retrieved for: {test_query}\n")

for idx in indices[0]:
    image_id = image_ids[idx]
    meta = image_id_to_meta.get(image_id, {})

    # Check if target model name appears in metadata (case-insensitive)
    meta_text = json.dumps(meta).lower()
    if target_model_name and target_model_name not in meta_text:
        continue  # Skip unrelated images

    img_path = meta.get("img_path")
    try:
        img = Image.open(img_path).convert("RGB")
        ipy_display(img)
        gemini_images.append(img)
    except Exception as e:
        print(f"❌ Could not load image: {img_path}: {e}")

if not gemini_images:
    print("⚠️ No relevant images found for this model.")
else:
    print(f"✅ Selected {len(gemini_images)} relevant images.")

# ✅ Retrieve matching text documents using LlamaIndex
retrieval_results = retriever.retrieve(test_query)
retrieved_texts = [res.text for res in retrieval_results]
document_context = "\n\n".join(retrieved_texts[:3])  # Top 3 results

# ✅ Ask Gemini using multimodal input (text + images)
model = genai.GenerativeModel("gemini-1.5-flash")

try:
    response = model.generate_content(
        [
            f"""
            Based on the following documents and images, answer this: {test_query}

            - Compare Hyundai models for long-distance travel.
            - Mention comfort, legroom, boot space, fuel efficiency, and infotainment.
            - Use car specs and images to support the comparison.

            DOCUMENTS:\n{document_context}
            """,
            *gemini_images
        ]
    )

    print("\n🎯 Gemini Multimodal Response:\n")
    print(response.text)

except Exception as e:
    print(f"❌ Gemini error: {e}")
```
